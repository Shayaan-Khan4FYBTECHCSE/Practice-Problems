```python
#DICTIONARY PROBLEMS

#Q1: Add key and value to a dictionary
my_dict={0:10,1:20}
my_dict[2]=30
print('Dictionary:',my_dict)

#Q2: Concatenate following dictionaries to create a new one
dic1={1:10,2:20}
dic2={3:30,6:60}
dic3={5:50,6:60}
new_dic={}
new_dic.update(dic1)
new_dic.update(dic2)
new_dic.update(dic3)
print('Dictionary no.1',dic1)
print('Dictionary no.2',dic2)
print('Dictionary no.3',dic3)
print('New dictionary',new_dic)

#Q3: Check if key is in dictionary
my_dict={101:'Amit',102:'Rahul',103:'Priya'}
key=int(input('Enter key to search:'))
if key in my_dict:
    print("Key exists in dictionary")
else:
    print("Key doesn't exists in dictionary")

#Q4: Use for loops to print keys alone, values alone and both keys and values
my_dict={101:'Amit',102:'Rahul',103:'Priya'}
print('Keys')
for key in my_dict.keys():
    print(key)
print()
print('Values')
for values in my_dict.values():
    print(values)
print()
print('Key : Value')
for k,v in my_dict.items():
    print(k,':',v)

#Q5: Prepare dictionay with keys from 1 to 15 with their values being there square
my_dict={}
for i in range(1,16):
    my_dict[i]=i**2
print('Dictionary:',my_dict)

#Q6: Sum all values of dictionary
my_dict={'a':10,'b':20,'c':30,'d':40}
total=0
for value in my_dict.values():
    total+=value
print('Sum of values:',total)

#Q7: Assign 5 elements in dictionary, Add and delete student information, Display student information
student_info={}
for i in range(5):
    roll_no=int(input('Enter roll no.:'))
    name=input('Enter name:')
    marks=int(input('Enter marks:'))
    student_info[roll_no]=[name,marks]
    
new_roll_no=int(input('Enter roll no.:'))
new_name=input('Enter name:')
new_marks=int(input('Enter marks:'))
student_info[new_roll_no]=[new_name,new_marks]

del_roll_no=int(input('Enter roll no. to delete:'))
del my_dict[del_roll_no]

print('Roll No.','Name','Marks',sep='\t')
for k, v in my_dict.items():
    print(k,end='\t')
    for n in v:
        print(n,end='\t')
    print()

#Q8: Convert lists into dictionary
list1=['name','panel','rollno']
list2=['ABC','B','34']
my_dict=dict(zip(list1,list2))
print('Dictionary:',my_dict)
```
