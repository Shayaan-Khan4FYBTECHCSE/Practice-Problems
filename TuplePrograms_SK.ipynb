```python
#TUPLE PROGRAMS

#Q1: 4th element from first and 4th element from last
numbers=(10,20,30,40,50,60,70,80,90)
print('Tuple:',numbers)
print('4th element from first:',numbers[3])
print('4th element from last:',numbers[-4])

#Q2: Check whethere element is in tuple or not
numbers=(10,20,30,40,50)
element=int(input('Enter element to search:'))
if element in numbers:
    print('Element exists in tuple')
else:
    print("Element doesn't exist in tuple")

#Q3: Convert list to tuple
numbers_list=[10,20,30,40,50]
numbers_tuple=tuple(numbers_list)
print('List:',numbers_list)
print('Tuple:'numbers_tuple)

#Q4: Find index of element in tuple
numbers=(10,20,30,40,50)
element=int(input('Enter element:'))
if element in numbers:
    print('Index of element:',numbers.index(element))
else:
    print('Element not found')

#Q5: Replace last value of tuples in a list to 100
numbers=[(10,20,40),(40,50,60),(70,80,90)]
new_list=[]
for i in numbers:
    new_tup=i[:-1]+(100,)
    new_list.append(new_tup)
print('Original list:',numbers)
print('New list:',new_list)

#Q6: Find max and min element of tuple
numbers=(25,10,45,5,60,30)
maximum=max(numbers)
minimum=min(numbers)
print('Tuple:',numbers)
print('Max element:',maximum)
print('Min element:',minimum)

#Q7: Swap last and first elements of a tuple
numbers=(10,20,30,40,50)
new_tup=(numbers[-1],)+numbers[1:-1]+(numbers[0])
print('Original tuple:',numbers)
print('New tuple:',new_tup)
```
